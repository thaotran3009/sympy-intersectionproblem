```python
import sympy as sp

# 1. sp.symbols: Khai bao bien x va tham so m tren tap so thuc
x, m = sp.symbols('x m', real=True)

# 2. sp.Eq: Thiet lap phuong trinh hoanh do giao diem: x^3 - 3x = m
f = x**3 - 3*x
eq = sp.Eq(f, m)
print("1. Phuong trinh:", eq)

# 3. sp.diff: Tinh dao ham f'(x) de khao sat cuc tri
f_prime = sp.diff(f, x)
print("2. Dao ham f'(x) =", f_prime)

# 4. sp.roots: Lay tu dien nghiem va so boi de loc cuc tri thuc su
roots_dict = sp.roots(f_prime, x)
print("3. Tap nghiem va so boi cua f'(x):", roots_dict)
crit_pts = [pt for pt, mult in roots_dict.items() if mult % 2 != 0]
print("   Cac diem cuc tri hop le:", crit_pts)

# 5. expr.subs: Tinh gia tri cuc dai va cuc tieu
y_cd = f.subs(x, -1)
y_ct = f.subs(x, 1)
print(f"4. Gia tri cuc dai: y_cd = {y_cd}, cuc tieu: y_ct = {y_ct}")
print(f"   => Dieu kien de duong thang y = m cat do thi tai 3 diem: {y_ct} < m < {y_cd}")

# 6. expr.subs: The tham so m = 0 vao phuong trinh
eq_m0 = eq.subs(m, 0)
print("5. Phuong trinh khi m = 0:", eq_m0)

# 7. sp.solve: Giai phuong trinh tim hoanh do giao diem chinh xac
nghiem_m0 = sp.solve(eq_m0, x)
print("6. Toa do hoanh do giao diem khi m = 0:", nghiem_m0)
```
